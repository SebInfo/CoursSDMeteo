```python
def bonjour():
    return "Bonjour !"
```

```python
def bonjour_aurevoir(fonction):
    print(fonction(), "Au revoir !", sep="\n")
```

```python
bonjour_aurevoir(bonjour)
```

```python
def mon_premier_decorateur(fonction):
    def bonjour():
        print("Bonjour !")
        fonction()
    return bonjour
```

```python
def fonction_decoree():
    print("Au revoir !")
```

```python
fonction_decoree = mon_premier_decorateur(fonction_decoree)
```

```python
fonction_decoree()
```
